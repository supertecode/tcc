```python
import pandas as pd
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

```python
df = pd.read_excel("amostra_sentimentos_modelo_novo_final.xlsx")
```

```python
# Dicionário de mapeamento
mapa = {
    'Positive': 1,
    'Neutral': 0,
    'Negative': -1
}

df['sentimento_real_num'] = df['Predição Correta'].map(mapa)
df['sentimento_pred_num'] = df['review_comment_message_sentiment'].map(mapa)
```

```python
# Classes reais e preditas
y_true = df['sentimento_real_num']                     # nome da coluna real
y_pred = df['sentimento_pred_num']     # nome da coluna de previsão

# Acurácia
accuracy = accuracy_score(y_true, y_pred)

# Precisão, recall e F1 para 3 classes
precision = precision_score(y_true, y_pred, average='macro')
recall    = recall_score(y_true, y_pred, average='macro')
f1        = f1_score(y_true, y_pred, average='macro')

print(f"Acurácia:  {accuracy:.4f}")
print(f"Precisão:  {precision:.4f}")
print(f"Recall:    {recall:.4f}")
print(f"F1-score:  {f1:.4f}")
```
